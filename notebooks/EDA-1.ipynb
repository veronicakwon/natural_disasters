## Fatality Rate (Total Deaths / Total Affected) distribution

```python
df['fatality_rate'] = df['Total Deaths'] / df['Total Affected']
df['fatality_rate']
```

```python
df['fatality_rate'].plot(kind='hist', bins=30, title='Fatality Rate Distribution', figsize=(10,5))
```

```python
df['fatality_rate'].describe()
df[df['fatality_rate'] > 10][['Disaster Type', 'Total Deaths', 'Total Affected', 'fatality_rate']].head(10)
```

## There appears to be some implausible fatality rates.
### Drop fatality rates that are greater than 1. 

```python
(df['fatality_rate'] > 1).sum()
```

```python
df['fatality_rate'] = df['fatality_rate'].clip(upper=1)
df['fatality_rate']
```

## Graph again

```python
df['fatality_rate'].plot(kind='hist', bins=30, title='Fatality Rate Distribution', figsize=(10,5))
```

## Distribution of economic losses over time (adjusted)

```python
reference_cpi = df.loc[df['Year'] == df['Year'].max()-1, 'CPI'].mean()

df['damages_adjusted'] = df["Total Damages ('000 US$)"] * (reference_cpi / df['CPI'])

df.groupby('Year')['damages_adjusted'].mean().plot(kind='line', title='Average Losses by Year (Adjusted for Inflation)', figsize=(10,5))
```
